Random Forest Classifier


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from imblearn.combine import SMOTEENN

# Step 1: Split original data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Step 2: Apply SMOTE only on training data
sm = SMOTEENN()
x_train_res, y_train_res = sm.fit_resample(x_train, y_train)

# Step 3: Train model on resampled data
model_smote_rf = RandomForestClassifier(
    n_estimators=100,
    criterion='gini',
    random_state=100,
    max_depth=6,
    min_samples_leaf=8
)

model_smote_rf.fit(x_train_res, y_train_res)

# Step 4: Predict on ORIGINAL test data
y_pred_smote_rf = model_smote_rf.predict(x_test)

# Step 5: Evaluation
print(classification_report(y_test, y_pred_smote_rf))
print(confusion_matrix(y_test, y_pred_smote_rf))
```

I applied SMOTEENN only on training data to avoid data leakage and ensure unbiased model evaluation.
